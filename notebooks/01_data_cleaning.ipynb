```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("Libraries loaded!")
```

```python
df = pd.read_csv('../data/Amazon Sale Report.csv', 
                 encoding='unicode_escape', 
                 low_memory=False)

print(f"Total rows: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum
```

```python
print(df.columns.tolist())
```

```python
df.drop(columns=['index', 'Unnamed: 22', 'promotion-ids', 
                 'fulfilled-by', 'ASIN'], inplace=True)

print(f"Columns remaining: {df.shape[1]}")
```

```python
df.dropna( subset= ['Amount'], inplace=True)
print(f"Rows remaining: {df.shape[0]}")
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month_name()
df['Month_Num'] = df['Date'].dt.month

print(df['Month'].value_counts())
```

```python
df.rename(columns={'Sales Channel ': 'Sales_Channels'}, inplace= True)
print("Done!")
```

```python
df.to_csv('../data/amazon_cleaned.csv', index=False)

print(f"Cleaned file saved!")
print(f"Final shape: {df.shape}")
```
